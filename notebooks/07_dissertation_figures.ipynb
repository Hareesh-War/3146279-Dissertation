# Figure 4.4 & 4.5: Phase 3b Training Curves

```python
def plot_training_curves(history, save_prefix='phase3b'):
    """Generate training accuracy and loss curves."""
    epochs = range(1, len(history['accuracy']) + 1)

    # Figure 4.4: Accuracy
    fig, ax = plt.subplots(figsize=(10, 5))
    ax.plot(epochs, history['accuracy'], 'b-', linewidth=2, label='Training Accuracy')
    ax.plot(epochs, history['val_accuracy'], 'r-', linewidth=2, label='Validation Accuracy')

    # Mark LR reduction point
    if 'lr' in history:
        lr_changes = []
        for i in range(1, len(history['lr'])):
            if history['lr'][i] != history['lr'][i-1]:
                lr_changes.append(i+1)
        for ep in lr_changes:
            ax.axvline(x=ep, color='gray', linestyle='--', alpha=0.7, label=f'LR Reduced (Epoch {ep})')

    ax.set_xlabel('Epoch')
    ax.set_ylabel('Accuracy')
    ax.set_title('Phase 3b: Training and Validation Accuracy')
    ax.legend(loc='lower right')
    ax.grid(True, alpha=0.3)
    ax.set_ylim(0, 1)
    fig.savefig(f'{FIGURE_DIR}/fig_4_4_{save_prefix}_accuracy.png')
    plt.show()
    print(f"Saved: {FIGURE_DIR}/fig_4_4_{save_prefix}_accuracy.png")

    # Figure 4.5: Loss
    fig, ax = plt.subplots(figsize=(10, 5))
    ax.plot(epochs, history['loss'], 'b-', linewidth=2, label='Training Loss')
    ax.plot(epochs, history['val_loss'], 'r-', linewidth=2, label='Validation Loss')

    if 'lr' in history:
        for ep in lr_changes:
            ax.axvline(x=ep, color='gray', linestyle='--', alpha=0.7, label=f'LR Reduced (Epoch {ep})')

    ax.set_xlabel('Epoch')
    ax.set_ylabel('Loss')
    ax.set_title('Phase 3b: Training and Validation Loss')
    ax.legend(loc='upper right')
    ax.grid(True, alpha=0.3)
    fig.savefig(f'{FIGURE_DIR}/fig_4_5_{save_prefix}_loss.png')
    plt.show()
    print(f"Saved: {FIGURE_DIR}/fig_4_5_{save_prefix}_loss.png")

plot_training_curves(phase3b_history)
```

# Figure 5.2: Per-Species F1-Score Bar Chart

```python
def plot_species_f1(metrics, save_name='fig_5_2_species_f1.png'):
    """Horizontal bar chart of per-species F1 scores, ranked."""
    species_data = metrics.get('per_species', metrics)

    # Sort by F1 score
    sorted_species = sorted(species_data.items(), key=lambda x: x[1]['f1'], reverse=False)

    names = [s[0] for s in sorted_species]
    f1_scores = [s[1]['f1'] * 100 for s in sorted_species]
    supports = [s[1]['support'] for s in sorted_species]

    # Color code by performance
    colors = []
    for f1 in f1_scores:
        if f1 >= 80:
            colors.append('#2ecc71')  # Green - good
        elif f1 >= 50:
            colors.append('#f39c12')  # Orange - moderate
        else:
            colors.append('#e74c3c')  # Red - poor

    fig, ax = plt.subplots(figsize=(12, 8))
    bars = ax.barh(names, f1_scores, color=colors, edgecolor='white', linewidth=0.5)

    # Add value labels
    for bar, f1, support in zip(bars, f1_scores, supports):
        if f1 > 0:
            ax.text(bar.get_width() + 1, bar.get_y() + bar.get_height()/2,
                    f'{f1:.1f}% (n={support})', va='center', fontsize=9)
        else:
            ax.text(2, bar.get_y() + bar.get_height()/2,
                    f'0.0% (n={support})', va='center', fontsize=9, color='gray')

    ax.set_xlabel('F1-Score (%)')
    ax.set_title('Phase 3b: Per-Species F1-Score (20 Species)')
    ax.axvline(x=75, color='blue', linestyle=':', alpha=0.5, label='75% Target')
    ax.set_xlim(0, 110)
    ax.legend(loc='lower right')
    ax.grid(True, alpha=0.2, axis='x')

    fig.savefig(f'{FIGURE_DIR}/{save_name}')
    plt.show()
    print(f"Saved: {FIGURE_DIR}/{save_name}")

plot_species_f1(phase3b_metrics)
```

# Figure 3.3: Sequential Transfer Learning Pipeline Diagram

```python
def plot_pipeline_diagram(save_name='fig_3_3_pipeline.png'):
    """Create the sequential transfer learning pipeline diagram."""
    fig, ax = plt.subplots(figsize=(14, 5))
    ax.set_xlim(0, 14)
    ax.set_ylim(0, 5)
    ax.axis('off')

    # Box positions and labels
    boxes = [
        (0.5, 2, 'ImageNet\n(1.2M images)\n1000 classes', '#3498db'),
        (3.0, 2, 'Phase 1\nDataset 1\n6 species\n1,380 imgs', '#2ecc71'),
        (5.5, 2, 'Phase 2\nDataset 2\n6 species\n1,094 imgs', '#27ae60'),
        (8.0, 2, 'Phase 3a\nDataset 4\n16 species\n3,494 imgs', '#e67e22'),
        (10.5, 2, 'Phase 3b\nDataset 5\n20 species\n2,503 imgs', '#e74c3c'),
    ]

    # Test set box
    test_box = (10.5, 0.3, 'Dataset 3\nIndependent Test\n9 species, ~900 imgs', '#9b59b6')

    for x, y, text, color in boxes:
        rect = matplotlib.patches.FancyBboxPatch(
            (x, y), 2.0, 2.2, boxstyle='round,pad=0.1',
            facecolor=color, alpha=0.3, edgecolor=color, linewidth=2
        )
        ax.add_patch(rect)
        ax.text(x + 1.0, y + 1.1, text, ha='center', va='center', fontsize=8, fontweight='bold')

    # Test set box
    x, y, text, color = test_box
    rect = matplotlib.patches.FancyBboxPatch(
        (x, y), 2.0, 1.3, boxstyle='round,pad=0.1',
        facecolor=color, alpha=0.3, edgecolor=color, linewidth=2, linestyle='--'
    )
    ax.add_patch(rect)
    ax.text(x + 1.0, y + 0.65, text, ha='center', va='center', fontsize=7, fontweight='bold')

    # Arrows between phases
    arrow_style = dict(arrowstyle='->', lw=2, color='#2c3e50')
    for i in range(len(boxes) - 1):
        x1 = boxes[i][0] + 2.0
        x2 = boxes[i+1][0]
        y_mid = boxes[i][1] + 1.1
        ax.annotate('', xy=(x2, y_mid), xytext=(x1, y_mid), arrowprops=arrow_style)

    # Arrow from Phase 3b to test set
    ax.annotate('', xy=(11.5, 1.6), xytext=(11.5, 2.0),
                arrowprops=dict(arrowstyle='->', lw=1.5, color='#9b59b6', linestyle='dashed'))

    # Labels above arrows
    labels = ['Pre-trained\nweights', 'Phase 1\nweights', 'Phase 2\nweights', 'Phase 3a\nweights']
    for i, label in enumerate(labels):
        x_mid = (boxes[i][0] + 2.0 + boxes[i+1][0]) / 2
        ax.text(x_mid, 3.5, label, ha='center', va='center', fontsize=7, color='#2c3e50')

    # Phase descriptions at bottom
    descs = ['General\nFeatures', 'Regional\nAdaptation', 'Endangered\nSpecialisation', 'Commercial\nValidation']
    for i, desc in enumerate(descs):
        ax.text(boxes[i+1][0] + 1.0, 4.5, desc, ha='center', va='center', fontsize=8,
                fontstyle='italic', color='#7f8c8d')

    ax.set_title('Sequential Transfer Learning Pipeline', fontsize=14, fontweight='bold', pad=20)

    fig.savefig(f'{FIGURE_DIR}/{save_name}')
    plt.show()
    print(f"Saved: {FIGURE_DIR}/{save_name}")

plot_pipeline_diagram()
```

# Figure 3.5: Progressive Layer Unfreezing Illustration

```python
def plot_unfreezing_strategy(save_name='fig_3_5_unfreezing.png'):
    """Illustrate the progressive unfreezing across phases."""
    phases = ['Phase 1', 'Phase 2', 'Phase 3a', 'Phase 3b']
    total_layers = 175
    frozen = [175, 175, 145, 155]
    unfrozen = [0, 0, 30, 20]
    lr = [1e-3, 1e-4, 5e-5, 5e-5]

    fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(12, 5))

    # Left: Stacked bar chart of frozen/unfrozen layers
    x = np.arange(len(phases))
    width = 0.5
    ax1.bar(x, frozen, width, label='Frozen Layers', color='#3498db', alpha=0.7)
    ax1.bar(x, unfrozen, width, bottom=frozen, label='Unfrozen Layers', color='#e74c3c', alpha=0.7)
    ax1.set_ylabel('Number of ResNet50 Layers')
    ax1.set_title('Progressive Layer Unfreezing')
    ax1.set_xticks(x)
    ax1.set_xticklabels(phases)
    ax1.legend()
    ax1.set_ylim(0, 200)
    for i, (f, u) in enumerate(zip(frozen, unfrozen)):
        if u > 0:
            ax1.text(i, f + u/2, f'{u}', ha='center', va='center', fontweight='bold', color='white')

    # Right: Learning rate progression
    ax2.semilogy(x, lr, 'ro-', linewidth=2, markersize=10)
    ax2.set_ylabel('Learning Rate (log scale)')
    ax2.set_title('Learning Rate Across Phases')
    ax2.set_xticks(x)
    ax2.set_xticklabels(phases)
    ax2.grid(True, alpha=0.3)
    for i, rate in enumerate(lr):
        ax2.text(i, rate * 1.5, f'{rate:.0e}', ha='center', fontsize=9)

    fig.suptitle('Progressive Unfreezing and Learning Rate Strategy', fontsize=13, fontweight='bold')
    plt.tight_layout()

    fig.savefig(f'{FIGURE_DIR}/{save_name}')
    plt.show()
    print(f"Saved: {FIGURE_DIR}/{save_name}")

plot_unfreezing_strategy()
```

```python
def plot_confusion_matrix_from_predictions(y_true, y_pred, class_names, save_name='fig_5_1_confusion_matrix.png'):
    """Generate confusion matrix heatmap from predictions."""
    from sklearn.metrics import confusion_matrix

    cm = confusion_matrix(y_true, y_pred)
    cm_normalized = cm.astype('float') / cm.sum(axis=1)[:, np.newaxis]

    fig, ax = plt.subplots(figsize=(16, 14))
    sns.heatmap(cm_normalized, annot=True, fmt='.2f', cmap='Blues',
                xticklabels=class_names, yticklabels=class_names, ax=ax,
                vmin=0, vmax=1, linewidths=0.5)
    ax.set_xlabel('Predicted Species')
    ax.set_ylabel('True Species')
    ax.set_title('Phase 3b: Normalised Confusion Matrix (20 Species)')
    plt.xticks(rotation=45, ha='right')
    plt.yticks(rotation=0)

    fig.savefig(f'{FIGURE_DIR}/{save_name}')
    plt.show()
    print(f"Saved: {FIGURE_DIR}/{save_name}")

# Load model and get predictions on the validation set:
# model = tf.keras.models.load_model('path/to/phase3b_model.keras')
# val_ds = tf.keras.utils.image_dataset_from_directory(...)
# y_true, y_pred = [], []
# for images, labels in val_ds:
#     predictions = model.predict(images)
#     y_pred.extend(np.argmax(predictions, axis=1))
#     y_true.extend(labels.numpy())
# plot_confusion_matrix_from_predictions(y_true, y_pred, class_names)

print("Confusion matrix function defined. Run with actual model predictions in Colab.")
```
